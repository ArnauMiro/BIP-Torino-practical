## Part B: Parametrization

> **A live demo of the NACA 4 digit parametrization**

Remember, NACAMPXX: 
* M is the maximum camber divided by 100
* P is the position of the maximum camber divided by 10
* XX is the thickness divided by 100

How does it work on the code:
```
m      maximum camber, PERCENT of chord      e.g. 4.0 for NACA 4412
p      position of max camber, FRACTION      e.g. 0.4 for NACA 4412
t      maximum thickness, PERCENT of chord   e.g. 12.0 for NACA 4412
```

````python
import itertools
import numpy as np
import matplotlib.pyplot as plt

import p1_ancillary as p1
````

### Thickness alone demo

````python
m = [2.0,2.0]   # Bounds for camber
p = [4.0,4.0]   # Bounds for position of max camber
t = [6.0,24.0] # Bounds for thickness
````

Thickness is nearly a pure scaling of the distribution

````python
mm = np.arange(m[0], m[1] + 1.0)
pp = np.arange(p[0], p[1] + 1.0)/10.
tt = np.arange(t[0], t[1] + 1.0)

# Generate Cartesian product of all mm, pp, and tt combinations
combinations = list(itertools.product(mm, pp, tt))

# Create grid; increased vertical figsize to fit 10 rows properly
nrow,ncol = 40,4
fig, ax = plt.subplots(nrow, ncol, figsize=(ncol*7, nrow*2))

# Iterate through axes and parameter combinations simultaneously
for axis, (m_val, p_val, t_val) in zip(ax.flat, combinations):
    p1.plot_airfoil(m_val, p_val, t_val, ax=axis)
    axis.set_title(p1.naca4_name(m_val, p_val, t_val))
    axis.set_axis_off()

# Hide any remaining empty subplots if combinations are fewer than 40
for axis in ax.flat[len(combinations):]:
    axis.set_visible(False)
````
